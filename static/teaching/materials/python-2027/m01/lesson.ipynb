Predict which object ends ahead and which has larger signed interval-average velocity. The records are synthetic.

```python
times_s = [0, 2, 4, 6, 8]
a_mm = [10, 14, 18, 22, 26]
b_mm = [0, 6, 12, 18, 24]
elapsed_s = times_s[-1] - times_s[0]
a_change_mm = a_mm[-1] - a_mm[0]
b_change_mm = b_mm[-1] - b_mm[0]
print("A (mm/s):", a_change_mm / elapsed_s)
print("B (mm/s):", b_change_mm / elapsed_s)
```
