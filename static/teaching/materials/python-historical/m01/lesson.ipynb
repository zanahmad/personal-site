# A record with units
Predict the temperature change and elapsed minutes before running. These data are synthetic; no cooling law is fitted.

```python
time_seconds = [0, 60, 120, 180, 240]
temperature_c = [80, 62, 48, 38, 31]
elapsed_minutes = (time_seconds[-1] - time_seconds[0]) / 60
change_c = temperature_c[-1] - temperature_c[0]
print("Elapsed (min):", elapsed_minutes)
print("Change (degrees C):", change_c)
print("Average change (degrees C/min):", change_c / elapsed_minutes)
```
